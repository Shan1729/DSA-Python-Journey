## Given a 1-indexed array of integers numbers that is already sorted in non-decreasing order, find two numbers such that they add up to a specific target number. Let these two numbers be numbers[index1] and numbers[index2] where 1 <= index1 < index2 <= numbers.length.
space.

## Return the indices of the two numbers index1 and index2, each incremented by one, as an integer array [index1, index2] of length 2.

## The tests are generated such that there is exactly one solution. You may not use the same element twice.

## Your solution must use only constant extra space.

## EXAMPLE 01: Input: numbers = [2,7,11,15], target = 9  
### 
Output: [1,2 
Explanation: The sum of 2 and 7 is 9. Therefore, index1 = 1, index2 = 2. We return [1, 2]

## Python version

```python
#Initialise left and right pointers for the sorted array.
left=0
right= len(numbers)-1

#While left is before right, find current sum.
while left<right:
    current_sum= numbers[left] + numbers[right]
    
    #Check current sum with target
    if current_sum == target:
        return [left+1 , right+1] #1-indexed form
        
    elif current_sum < target:
        left+=1                   #Increment left

    else:
        right-=1                  #Decrement right

return -1
```

| 3 | 0 | 1 | 9 | Target Found → Return [1,2] |

## Python function version

```python
def two_sum_sorted(numbers, target):
    #Initialise left and right pointers for the sorted array.
    left=0
    right= len(numbers)-1

#While left is before right, find current sum.
    while left<right:
        current_sum= numbers[left] + numbers[right]
    
    #Check current sum with target
        if current_sum == target:
            return [left+1 , right+1] #1-indexed form return means end it right now with the answers
        
        elif current_sum < target:
            left+=1                   #Increment left

        else:
            right-=1                  #Decrement right

    return -1
    
```

## Problem 
You are given a sorted array of integers and a target 
Return all pairs of values whose sum equals the targe .
For now, assu e:

target 10

## Dry Run — All Pairs With Target Sum

**Input:** `numbers = [1, 2, 3, 4, 5, 6, 7, 8]`  
**Target:** `10`

| Step | L | R | Pair | Current Sum | Action |
|---:|---:|---:|---|---:|---|
| 1 | 0 | 7 | (1, 8) | 9 | Sum < Target → Move L → L++ |
| 2 | 1 | 7 | (2, 8) | 10 | Found → Store (2, 8) → Move both |
| 3 | 2 | 6 | (3, 7) | 10 | Found → Store (3, 7) → Move both |
| 4 | 3 | 5 | (4, 6) | 10 | Found → Store (4, 6) → Move both |
| 5 | 4 | 4 | — | — | **Terminate: L == R** |

**Final Answer:** `[(2, 8), (3, 7), (4, 6)]`

# Python Function Version

```python
def two_sum_all_pairs(nums , target):

    #Initialise left and right and declare answer array to store all pairs.
    left= 0 
    right= len(nums)-1
    ans_arr= []

    # While left is less than right find current sum.
    while left<right:
        curr_sum = nums[left] + nums[right]

        if curr_sum == target:                #IF current sum equals target append the pair in 1-indexed form
            ans_arr.append([left+1,right+1])    #move left pointer by 1 and move right pointer by 1 in respective directions
            left+=1
            right-=1

        elif curr_sum < target:               #If current sum is less than target move left pointer to the right by 1.
            left+=1

        else:
            right-=1                          #If current sum is greaer than target move right pointer to the left by 1.

    return ans_arr
```
