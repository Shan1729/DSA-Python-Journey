# LC 125- Valid Palindrome (EASY) 

A phrase is a palindrome if, after converting all uppercase letters into lowercase letters and removing all non-alphanumeric characters, it reads the same forward and backward. Alphanumeric characters include letters and numbers.

Given a string s, return true if it is a palindrome, or false otherwise.

 

Example 1:

Input: s = "A man, a plan, a canal: Panama"
Output: true
Explanation: "amanaplanacanalpanama" is a palindrome.
Example 2:

Input: s = "race a car"
Output: false
Explanation: "raceacar" is not a palindrome.
Example 3:

Input: s = " "
Output: true
Explanation: s is an empty string "" after removing non-alphanumeric characters.
Since an empty string reads the same forward and backward, it is a palindrome.
 

Constraints:

1 <= s.length <= 2 * 105
s consists only of printable ASCII characters.

## Dry Run ¶

**Input:**  
`s = "A man, a plan, a canal: Panama"`

| Step | Left | Right | Left Char | Right Char | Decision |
|:---:|:---:|:---:|:---:|:---:|:---|
| 1 | 0 | 29 | `A` | `a` | Both valid → Compare → Equal → Move Both |
| 2 | 1 | 28 | ` ` | `m` | Left invalid → Move Left |
| 3 | 2 | 28 | `m` | `m` | Both valid → Compare → Equal → Move Both |
| 4 | 3 | 27 | `a` | `a` | Both valid → Compare → Equal → Move Both |
| 5 | 4 | 26 | `n` | `n` | Both valid → Compare → Equal → Move Both |
| 6 | 5 | 25 | `,` | `a` | Left invalid → Move Left |
| 7 | 6 | 25 | ` ` | `a` | Left invalid → Move Left |
| 8 | 7 | 25 | `a` | `a` | Both valid → Compare → Equal → Move Both |
| ... | ... | ... | ... | ... | ... |
| Final | `left >= right` | | | | All valid characters matched → **Return True** |

### Key Observation

**Clean Left → Clean Right → Compare → Move Both**

- Invalid Left → `left += 1`
- Invalid Right → `right -= 1`
- Both Valid → Compare after case normalization
- Mismatch → **Return False**
- Pointers meet/cross → **Return True**

## Final Code

```python
def is_palindrome(s):
    left= 0
    right= len(s) -1
    
    while left<right:             #Left and Right must not cross
        while left<right and not s[left].isalnum():   #Removing all left invalid
            left += 1

        while left<right and not s[right].isalnum():  #Removing all right invalid
            right-= 1

        if s[left].lower() != s[right].lower():       #Compare valids
            return False                       

        left+= 1      
        right-= 1                                     #Move pointers

    return True
```

```python
s1= "A man, a plan, a canal: Panama"
print(is_palindrome(s1))
```

```python
s2= "raCeCar"
print(is_palindrome(s2))
```

```python
s3= ""
print(is_palindrome(s3))
```
