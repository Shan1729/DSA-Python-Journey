# LEETCODE 643- Maximum Average Subarray 1 (EASY)

You are given an integer array nums consisting of n elements, and an integer k.

Find a contiguous subarray whose length is equal to k that has the maximum average value and return this value. Any answer with a calculation error less than 10-5 will be accepted.

 

Example 1:

Input: nums = [1,12,-5,-6,50,3], k = 4
Output: 12.75000
Explanation: Maximum average is (12 - 5 - 6 + 50) / 4 = 51 / 4 = 12.75
Example 2:

Input: nums = [5], k = 1
Output: 5.00000
 

Constraints:

n == nums.length
1 <= k <= n <= 105
-104 <= nums[i] <= 104

## Dry Run

### Input
`nums = [1, 12, -5, -6, 50, 3]`  
`k = 4`

| Step | Left | Right | Current Sum | Decision |
|------|------|-------|-------------|----------|
| Initial | 0 | 3 | 2 | First window → Set `max_sum = 2` |
| 1 | 1 | 4 | 51 | `51 > 2` → Update `max_sum = 51` |
| 2 | 2 | 5 | 42 | `42 < 51` → Keep `max_sum = 51` |

### Window Movement

```text
Initial Window:
[1, 12, -5, -6]  50  3
 ↑           ↑
left       right

After Step 1:
 1  [12, -5, -6, 50]  3
     ↑            ↑
    left        right

After Step 2:
 1   12  [-5, -6, 50, 3]
         ↑           ↑
        left       right

## Python Version

````python
def max_subarray_avg(nums, k):
    #Initialise pointers
    left= 0
    right= k-1
    current_sum = sum(nums[ : k])    #Sum of first k elements
    max_sum = current_sum            #Initial max sum is current sum
    
    # Declare the while condition
    while right < len(nums)-1:   #This is because the last element will get added to the final sum when right points to n-2th element.

        #subtract left element
        current_sum -= nums[left]
        
        #Move both pointers by 1 in the same direction
        left += 1
        right += 1

        # Add the incoming right element
        current_sum += nums[right]
        
        #Compare new sum with max sum
        if current_sum > max_sum:
             max_sum = current_sum

    return max_sum/k

````

````python
nums = [1,2,3,-40,49,1,2]
k1 = 4
x1= max_subarray_avg(nums, k1)
print(x1)
````
