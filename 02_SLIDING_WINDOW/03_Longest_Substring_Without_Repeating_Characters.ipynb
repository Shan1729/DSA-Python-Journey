# LEETCODE 03 : Longest Substring without repeating characters  (MEDIUM)


## Given a string s, find the length of the longest substring without duplicate characters.

 

Example 1:

Input: s = "abcabcbb"
Output: 3
Explanation: The answer is "abc", with the length of 3. Note that "bca" and "cab" are also correct answers.
Example 2:

Input: s = "bbbbb"
Output: 1
Explanation: The answer is "b", with the length of 1.
Example 3:

Input: s = "pwwkew"
Output: 3
Explanation: The answer is "wke", with the length of 3.
Notice that the answer must be a substring, "pwke" is a subsequence and not a substring.

# Dry Run

## Input

s = "abcabcbb"

### Dry Run

| Step | Left | Right | Current Character | Window | Current Length | Max Length | Decision |
|------|------|-------|-------------------|--------|----------------|------------|----------|
| Initial | 0 | 0 | a | {a} | 1 | 1 | First character → Set max_length = 1 |
| 1 | 0 | 1 | b | {a,b} | 2 | 2 | 2 > 1 → Update max_length = 2 |
| 2 | 0 | 2 | c | {a,b,c} | 3 | 3 | 3 > 2 → Update max_length = 3 |
| 3 | 1 | 3 | a | {b,c,a} | 3 | 3 | Duplicate a → Remove characters from left until a is removed |
| 4 | 2 | 4 | b | {c,a,b} | 3 | 3 | Duplicate b → Remove characters from left until b is removed |
| 5 | 3 | 5 | c | {a,b,c} | 3 | 3 | Duplicate c → Remove characters from left until c is removed |
| 6 | 5 | 6 | b | {c,b} | 2 | 3 | Duplicate b → Shrink window |
| 7 | 7 | 7 | b | {b} | 1 | 3 | Duplicate b → Shrink window |

### Final Answer

Maximum length = 3

Longest substring = "abc"

## Python Version

```python
def longest_substring_no_duplicates(s):
    left = 0             #Initialise left at 0th index
    max_length = 0       #This will be the answer initialise to 0
    window = set()       #Declare an emty set for the window (no duplicates means set is to be used

    #For loop for the right pointer as it iterates over every element.
    for right in range(0, len(s)):

        while s[right] in window:      #Only keep removing s[left] till no duplicate is remaining.
            window.remove(s[left])
            left += 1                     #keep moving pointer by 1 until duplicate is removed

        window.add(s[right])              #add s[right] to the window
                    
        current_length = right - left + 1   #current length of the window is right-left+1.

        max_length = max(max_length , current_length)      #max_length will be maximum length from comparison.

    return max_length
    
```
