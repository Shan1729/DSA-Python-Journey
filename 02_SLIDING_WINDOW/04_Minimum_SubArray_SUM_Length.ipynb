# Leetcode 209 Minimum Length Of SubArray (Medium)
Given an array of positive integers nums and a positive integer target, return the minimal length of a subarray whose sum is greater than or equal to target. If there is no such subarray, return 0 instead.

 

Example 1:

Input: target = 7, nums = [2,3,1,2,4,3]
Output: 2
Explanation: The subarray [4,3] has the minimal length under the problem constraint.
Example 2:

Input: target = 4, nums = [1,4,4]
Output: 1
Example 3:

Input: target = 11, nums = [1,1,1,1,1,1,1,1]
Output: 0
 

# Dry Run

## Input

target = 7
nums = [2, 3, 1, 2, 4, 3]

### Dry Run

| Step | Left | Right | Current Sum | Window | Current Length | Min Length | Decision |
|------|------|-------|-------------|--------|----------------|------------|----------|
| Initial | 0 | 0 | 2 | [2] | — | ∞ | Sum < 7 → Expand |
| 1 | 0 | 1 | 5 | [2,3] | — | ∞ | Sum < 7 → Expand |
| 2 | 0 | 2 | 6 | [2,3,1] | — | ∞ | Sum < 7 → Expand |
| 3 | 0 | 3 | 8 | [2,3,1,2] | 4 | 4 | Sum ≥ 7 → Record length, then shrink |
| 4 | 1 | 3 | 6 | [3,1,2] | — | 4 | Sum < 7 → Stop shrinking |
| 5 | 1 | 4 | 10 | [3,1,2,4] | 4 | 4 | Sum ≥ 7 → Record length, then shrink |
| 6 | 2 | 4 | 7 | [1,2,4] | 3 | 3 | Sum ≥ 7 → Update min_length = 3, then shrink |
| 7 | 3 | 4 | 6 | [2,4] | — | 3 | Sum < 7 → Stop shrinking |
| 8 | 3 | 5 | 9 | [2,4,3] | 3 | 3 | Sum ≥ 7 → Record length, then shrink |
| 9 | 4 | 5 | 7 | [4,3] | 2 | 2 | Sum ≥ 7 → Update min_length = 2, then shrink |
| 10 | 5 | 5 | 3 | [3] | — | 2 | Sum < 7 → Stop shrinking |

### Final Answer

Minimum length = 2

Shortest subarray = [4, 3]

# PSEUDOCODE

START

left = 0
minimum_length = +infinity
current_sum = 0

FOR right from 0 to len(nums) - 1:

    current_sum += nums[right]

    WHILE current_sum >= target:

        current_length = right - left + 1

        minimum_length = MIN(minimum_length, current_length)

        current_sum -= nums[left]

        left += 1

IF minimum_length == +infinity:

    RETURN 0

ELSE:

    RETURN minimum_length

# Python Version

```python
def min_subarray_sum_length( nums , target):
    #Initialise left and right pointers , min_length , current_sum 
    left = 0
    min_length = float('inf')
    current_sum = 0

    for right in range(len(nums)):
        #Expand window from right
        current_sum += nums[right]

        #Check for valid window
        while current_sum >= target:

            #record current_length and update minimum length
            current_length = right-left+1
            min_length = min(min_length , current_length)

            #Shrink the window to look for lesser length of valid window
            current_sum -= nums[left]
            left += 1
            
    #Handle edge case
    if min_length == float('inf'):
        return 0

    else:
        return min_length
```
