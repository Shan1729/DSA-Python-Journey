# Leetcode 01 TWO SUM EASY

## You are given an array of integers nums and an integer target, return indices of the two numbers such that they add up to target.

You may assume that each input would have exactly one solution, and you may not use the same element twice.

You can return the answer in any order.

 

Example 1:

Input: nums = [2,7,11,15], target = 9
Output: [0,1]
Explanation: Because nums[0] + nums[1] == 9, we return [0, 1].
Example 2:

Input: nums = [3,2,4], target = 6
Output: [1,2]
Example 3:

Input: nums = [3,3], target = 6
Output: [0,1]
 

Constraints:

2 <= nums.length <= 104
-109 <= nums[i] <= 109
-109 <= target <= 109
Only one valid answer exists.
 

# Dry Run

## Input

target = 9
nums = [2, 7, 11, 15]

### Dry Run

| Step | Index | Current Number | Complement | Hash Map | Decision |
|------|-------|----------------|------------|----------|----------|
| Initial | 0 | 2 | 7 | {} | 7 not found → Store 2 → 0 |
| 1 | 1 | 7 | 2 | {2: 0} | 2 found at index 0 → Return [0, 1] |

### Final Answer

Indices = [0, 1]

Two numbers = [2, 7]

2 + 7 = 9

## Python Version

```python
def two_sum( nums , target):
    #Initialise the hashmap where numbers and their indexes will be stored.
    map = {}

    #We need to check every index,number in nums list
    for index,num in enumerate(nums):
        #Calculate complement
        complement = target - num

        #check if complement exists or not in hashmap
        if complement in map:
            return ( map[complement] , index)    #If complement exists in hashmap, return its value ( position/index) along with current index.

        else: 
            map[num]= index
```

```python
nums = [2,7,11,15]
target = 9
print(two_sum( nums , target))
```
