# Leetcode 49 Group Anagrams Medium
Given an array of strings strs, group the anagrams together. You can return the answer in any order.

 

Example 1:

Input: strs = ["eat","tea","tan","ate","nat","bat"]

Output: [["bat"],["nat","tan"],["ate","eat","tea"]]

Explanation:

There is no string in strs that can be rearranged to form "bat".
The strings "nat" and "tan" are anagrams as they can be rearranged to form each other.
The strings "ate", "eat", and "tea" are anagrams as they can be rearranged to form each other.
Example 2:

Input: strs = [""]

Output: [[""]]

Example 3:

Input: strs = ["a"]

Output: [["a"]]

 

# Dry Run

## Input

strs = ["eat", "tea", "tan", "ate", "nat", "bat"]

### Dry Run

| Step | Word | Sorted Key | HashMap | Decision |
|------|------|------------|---------|----------|
| Initial | eat | aet | {} | Key not present → Create `aet: ["eat"]` |
| 1 | tea | aet | {"aet": ["eat"]} | Key exists → Append "tea" |
| 2 | tan | ant | {"aet": ["eat", "tea"]} | Key not present → Create `ant: ["tan"]` |
| 3 | ate | aet | {"aet": ["eat", "tea"], "ant": ["tan"]} | Key exists → Append "ate" |
| 4 | nat | ant | {"aet": ["eat", "tea", "ate"], "ant": ["tan"]} | Key exists → Append "nat" |
| 5 | bat | abt | {"aet": ["eat", "tea", "ate"], "ant": ["tan", "nat"]} | Key not present → Create `abt: ["bat"]` |

### Final HashMap

{
    "aet": ["eat", "tea", "ate"],
    "ant": ["tan", "nat"],
    "abt": ["bat"]
}

### Final Answer

[
    ["eat", "tea", "ate"],
    ["tan", "nat"],
    ["bat"]
]

## Lets write a PSEUDOCODE for this

## PYTHON VERSION

```python
def anagrams( strs ):
    group_anagrams = {}

    for word in strs:
        key = "".join(sorted(word))

        if key in group_anagrams:
            group_anagrams[key].append(word)
        else:
            group_anagrams[key] = [word]   #create new  key and store list form of word

    return list(group_anagrams.values())
```

```python
strs = ["eat" , "tan" , "nat" , "bat" , "ate"]
print(anagrams(strs))
```
